# Model building

```python
# Subset the data for inference
# Create bins (quartiles) based on GA_delivery values
full_data_pca['GA_bin'] = pd.qcut(full_data_pca['GA_delivery'], q=4, duplicates='drop')
# Stratified sampling: pick 2 samples per GA bin for inference
inference_samples_pca = (
    full_data_pca.groupby('GA_bin', group_keys=False)
    .sample(n=2, random_state=42)
)
# Remove inference samples from the original dataset and drop helper bin column
full_data_pca2 = full_data_pca.drop(inference_samples_pca.index).drop(columns='GA_bin')

full_data_percentile['GA_bin'] = pd.qcut(full_data_percentile['GA_delivery'], q=4, duplicates='drop')
inference_samples_percentile = (
    full_data_percentile.groupby('GA_bin', group_keys=False)
    .sample(n=2, random_state=42)
)
full_data_percentile2 = full_data_percentile.drop(inference_samples_percentile.index).drop(columns='GA_bin')

full_data_both['GA_bin'] = pd.qcut(full_data_both['GA_delivery'], q=4, duplicates='drop')
inference_samples_both = (
    full_data_both.groupby('GA_bin', group_keys=False)
    .sample(n=5, random_state=42)
)
full_data_both2 = full_data_both.drop(inference_samples_both.index).drop(columns='GA_bin')
```
