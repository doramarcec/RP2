```python
import pandas as pd
import numpy as np
```

```python
# Import the data
preterm_data = pd.read_csv("../output/tarca_et_al_2021/GSE150167-term+preterm/full_data_merged.csv")
term_data = pd.read_csv("../output/tarca_et_al_2021/GSE206454-term-only/full_data_merged.csv")
```

# Preprocessing

```python
# Add GEO accessions
preterm_data['GEO_accession'] = 'GSE150167'
term_data['GEO_accession'] = 'GSE206454'

# Add the Group column to the term data and label as controls
term_data['Group'] = 'Control'
```

```python
# Are all the Organism values the same and coming from human samples? Yes, they are. 
#term_data.value_counts('Organism_x')
#term_data.value_counts('Organism_y')
```

```python
# Remove redundant columns 
term_data = term_data.drop(['Organism_x', 'Organism_y', 'ID', 'ID_REF', 'SPOT_ID', 'SeqIdVersion', 'Type', 'Units'], axis = 1)
preterm_data = preterm_data.drop(['ID', 'PT_LIST', 'Sample_channel_count', 'Tissue', 'DiagnosisGA', 'SPOT_ID'], axis = 1)
preterm_data
```

```python
# Harmonize the column names
preterm_data = preterm_data.rename(columns={'ENTREZ_GENE_ID_LIST': 'EntrezGeneID'})
term_data = term_data.rename(columns={'SomaId': 'ID_REF'})
```

```python
# Add GA_delivery column and fill with NaNs
term_data['GA_delivery'] = 'NaN'
term_data
```

```python
# Filter the term dataset to only contain entrez gene symbols present in preterm dataset
filtered_term_data = term_data[term_data['EntrezGeneSymbol'].isin(preterm_data['EntrezGeneSymbol'])]
filtered_term_data
```

# Normalization

```python
# Restructure the datasets
filtered_term_data2 = filtered_term_data[['Accession', 'Target', 'Log2RFU']]
filtered_term_data3 = filtered_term_data2.drop_duplicates(subset = ['Accession', 'Target'], keep='last') # 753588 to 565710 (or 417276 if False)
filtered_term_data4 = filtered_term_data3.pivot(index = 'Accession', columns = 'Target', values = 'Log2RFU')
filtered_term_data4
```

```python
# Calculate a rowsum (total intensity in each sample)
rowsums = filtered_term_data4.sum(axis = 1)
rowsums

# Identify the maximum rowsum (maximum total intensity across samples)
max_total = rowsums.max()
max_total

# Normalise the data
term_normalised = filtered_term_data4.div(rowsums, axis = 0).mul(max_total) # normalized_sample = (sample / rowsum) * max_total
term_normalised

# Log2 transform the data to stabilise the variance
term_log2 = np.log2(term_normalised + 1e-6)
term_log2
```

```python
preterm_data2 = preterm_data[['Accession', 'Target', 'Log2RFU']]
preterm_data3 = preterm_data2.pivot(index = 'Accession', columns = 'Target', values = 'Log2RFU')
preterm_data3
```

> Data was log2 transformed before applying machine learning and differential abundance analyses ([Tarca et al., 2021](https://www.sciencedirect.com/science/article/pii/S266637912100166X?via%3Dihub#sec5)).

```python
# Calculate a rowsum (total intensity in each sample)
rowsums = preterm_data3.sum(axis = 1)
rowsums

# Identify the maximum rowsum (maximum total intensity across samples)
max_total = rowsums.max()
max_total

# Normalise the data
preterm_normalised = preterm_data3.div(rowsums, axis = 0).mul(max_total) # normalized_sample = (sample / rowsum) * max_total
preterm_normalised

# Log2 transform the data to stabilise the variance
#preterm_log2 = np.log2(preterm_normalised + 1e-6)
#preterm_log2
```

```python
# Combine the term dataset with its metadata
term_log2_long = term_log2.reset_index().melt(id_vars='Accession', var_name='Target', value_name='Log2RFU')
term_log2_long

# Remove the log2RFU col from the filtered term data to be able to merge
filtered_term_data = filtered_term_data.drop(columns = ['Log2RFU']) # hahaha ha ha... 

term_log2_full = pd.merge(term_log2_long, filtered_term_data, on = ['Accession', 'Target'], how = 'inner')
term_log2_full
```

```python
preterm_data
```

```python
# Combine the preterm dataset with its metadata
preterm_normalised_long = preterm_normalised.reset_index().melt(id_vars='Accession', var_name='Target', value_name='Log2RFU')
preterm_normalised_long

# Remove the log2RFU col from the filtered term data to be able to merge
preterm_data = preterm_data.drop(columns = ['Log2RFU']) # hahaha ha ha... 

preterm_norm_full = pd.merge(preterm_normalised_long, preterm_data, on = ['Accession', 'Target'], how = 'inner')
preterm_norm_full
```

```python
# Concatenate the two datasets
merged_data = pd.concat([preterm_norm_full, term_log2_full], ignore_index=True)
merged_data
```

```python
# Export the merged dataset
merged_data.to_csv("../output/tarca_et_al_2021/merged_normalised_datasets_for_modelling.csv", index=False) 
```
